## 3. Data Transformation and Analysis

After cleaning, the datasets were transformed to support the main business questions.

A DayOfWeek variable was created to examine weekly activity patterns. Average daily steps were calculated for each user, and users were grouped into project-defined activity levels:

- Low Activity: fewer than 5,000 average daily steps
- Moderate Activity: 5,000–9,999 average daily steps
- High Activity: 10,000 or more average daily steps

Additional analysis examined activity intensity, the relationship between daily steps and calories burned, and sleep duration across the week.

The activity and sleep datasets were also merged by user ID and date to explore relationships between activity and sleep behavior.

```python
user_avg_steps = daily_activity.groupby("Id")["TotalSteps"].mean()

print(user_avg_steps.describe())
```

```python
user_activity = user_avg_steps.reset_index()
user_activity.columns = ["Id", "AverageDailySteps"]

def activity_level(steps):
    if steps < 5000:
        return "Low activity"
    elif steps < 10000:
        return "Moderate activity"
    else:
        return "High activity"

user_activity["ActivityLevel"] = user_activity[
    "AverageDailySteps"
].apply(activity_level)

print(user_activity["ActivityLevel"].value_counts())

print("\nPercentage of users:")
print(
    user_activity["ActivityLevel"]
    .value_counts(normalize=True)
    .mul(100)
    .round(1)
)
```

```python
user_summary = daily_activity.groupby("Id").agg(
    AverageDailySteps=("TotalSteps", "mean"),
    DaysRecorded=("ActivityDate", "count")
).reset_index()

user_summary["ActivityLevel"] = user_summary[
    "AverageDailySteps"
].apply(activity_level)

print(
    user_summary.groupby("ActivityLevel")["DaysRecorded"]
    .agg(["count", "mean", "median", "min", "max"])
)
```

## 4. Key Findings

The analysis identified five main findings:

1. **Weekly activity patterns:** Users averaged approximately 7,377 steps per day. Saturday had the highest average daily steps at 7,752, while Sunday had the lowest at 6,607.

2. **Activity levels:** Based on the project-defined activity segments, 48.6% of users were moderately active, 31.4% were low activity, and 20.0% were highly active. Overall, 80% of users were in the low or moderate activity groups.

3. **Activity intensity:** Light activity dominated recorded active time, averaging 188.1 minutes per day compared with 19.9 very-active minutes and 13.6 fairly-active minutes.

4. **Steps and calories:** Daily steps had a moderate positive correlation with calories burned (r = 0.58), indicating that higher-step days generally tended to be associated with greater calorie expenditure.

5. **Sleep patterns:** Among the 24 users with available sleep data, average sleep duration was approximately 6.99 hours per night. Sunday had the highest average sleep duration at 7.55 hours, while Thursday had the lowest at 6.69 hours.

## 5. Business Recommendations

Based on the analysis, three recommendations are proposed for Bellabeat:

### 1. Personalize Achievable Movement Goals
Bellabeat could provide personalized daily movement goals based on users' activity levels. Since 80% of users in this analysis were classified as low or moderately active and light activity represented the largest portion of recorded active time, the app could encourage walking, short movement breaks, and gradual increases in daily activity.

### 2. Use Day-Specific Engagement
Bellabeat could use personalized reminders, challenges, and motivational messages based on users' activity patterns across the week. Rather than applying the same strategy to every user, engagement could adapt to each user's individual behavior.

### 3. Integrate Activity and Sleep Insights
Bellabeat could combine activity and sleep information to provide personalized weekly wellness summaries. This could help users better understand their movement and sleep patterns together.

## Conclusion

This analysis identified meaningful patterns in activity levels, activity intensity, calorie expenditure, and sleep behavior that could help inform Bellabeat's marketing and user-engagement strategy.

However, the findings are exploratory because the dataset represents a relatively small sample of Fitbit users rather than Bellabeat customers and was collected in 2016. Bellabeat should validate these patterns using larger, more recent data from its own users before making major marketing decisions.
